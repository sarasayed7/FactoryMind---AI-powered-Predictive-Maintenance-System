```python
def temperature_validation(value):
    # missing value
    if value is None:
        return f'{value} is missing or None'

    # bool
    elif type(value) == bool:
        return f'{value} is not a valid number'

    # is int or float
    elif type(value) == int or type(value) == float:
        return float(value)

    # string
    elif type(value) == str:
        try:
            return float(value)
        except ValueError:
            return f'{value} is not a valid number'
    else:
        return f'{value} is not a valid number'
```

# Full validation Pipeline

**Validate Sensor Rules**

```python
import math

def validate_limit(value):
    # None means no boundary restriction
    if value is None:
        return True

    # Reject Boolean values
    if isinstance(value, bool):
        return False

    # Reject values that are not integers or floats
    if not isinstance(value, (int, float)):
        return False

    # Reject NaN and infinity
    if not math.isfinite(value):
        return False

    return True
```

```python
test_values = [
    100,
    -50.5,
    None,
    "100",
    True,
    float("nan"),
    float("inf"),
    [100]
]

for value in test_values:
    print(value, "->", validate_limit(value))
```

```python
def validate_sensor_rules(sensor_rules):
    errors = []

    # empty sensor_rules = {}
    if not sensor_rules:
        errors.append(f'Sensor rules can not be empty')

    # 1. Iterate through sensor rules
    for field, rules in sensor_rules.items():
        # empty fields--> temp = {}
        if not rules:
            errors.append(f'{field} cannot be empty')
        # 2. Retrieve minimum and maximum values
        min_value = rules.get('min_value')
        max_value = rules.get('max_value')

        if not validate_limit(min_value) or not validate_limit(max_value):
            errors.append(f'{field} is a {type(field)}')
            continue

        # 3. Check for contradictory limits
        if min_value is not None and max_value is not None and min_value > max_value:
            # 4. Append any configuration errors
            errors.append(f'{field}: {min_value} cannot exceed {max_value}')

    is_valid = len(errors) == 0

    # 5. Return is_valid and errors
    return {
        "is_valid": is_valid,
        "errors": errors
    }
```

**Validate Sensor Data**

```python
import math
def validate_sensor_data(sensor, sensor_rules):
    errors = []
    validated_data = {}

    for field, rules in sensor_rules.items():

        # Step 1: Get sensor value
        value = sensor.get(field)

        # Step 2: Check if missing
        # Hint: What should happen if the value is None?
        if value is None:
            errors.append(f'{field} is missing or None')
            continue

        # Step 3: Validate and convert to float
        # Hint: How will you detect failed conversion?

        elif type(value) == bool:
            errors.append(f'{field} is a {type(field)}')
            continue

        # is int or float
        elif type(value) == int or type(value) == float:
            temp = float(value)

        # string
        elif type(value) == str:
            try:
                temp = float(value)
            except ValueError:
                errors.append(f'{field} is a {type(field)}')
                continue
        else:
            errors.append(f'{field} is not a valid number')
            continue

        # finite number validation
        if math.isfinite(temp) == False:
            errors.append(f'{field} is not a finite number')
            continue

        # Step 4: Apply min/max validation
        # Hint: You already wrote range_validation()
        min_value = rules.get('min_value')
        max_value = rules.get('max_value')

        if min_value is not None and max_value is not None and min_value > max_value:
            errors.append('Min value cannot be greater than max value')
            continue

        if min_value is not None and temp < min_value:
            errors.append(f'{field}: {value} is below minimum {min_value}')
            continue

        if max_value is not None and temp > max_value:
            errors.append(f'{field}: {value} exceeds maximum {max_value}')
            continue


        # Step 5: Store successfully validated value
        validated_data[field] = temp

    # Step 6: Determine is_valid
    is_valid = len(errors) == 0

    return {
        "is_valid": is_valid,
        "data": validated_data,
        "errors": errors
    }
```
